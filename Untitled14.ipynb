```python
!pip install -q transformers accelerate bitsandbytes safetensors torch torchvision

from transformers import AutoProcessor, AutoModelForVision2Seq
import torch
from PIL import Image
import requests
from io import BytesIO

# Load processor and model from Hugging Face
model_id = "llava-hf/llava-1.5-7b-hf"

processor = AutoProcessor.from_pretrained(model_id)
model = AutoModelForVision2Seq.from_pretrained(
    model_id,
    torch_dtype=torch.float16,
    low_cpu_mem_usage=True,
    device_map="auto"
)
```

```python
from google.colab import files
from PIL import Image
import torch

# Step 1: Upload new image every time
uploaded = files.upload()
image_path = list(uploaded.keys())[0]
image = Image.open(image_path)

# Step 2: Prompt
prompt = """<|user|>
<image>
Please analyze this image and tell me whether the chicken feed shown here is fresh or not.
Explain your reasoning based on texture, color, and composition.
<|end|>
<|assistant|>"""

# Step 3: Process input
inputs = processor(prompt, image, return_tensors="pt").to(
    "cuda" if torch.cuda.is_available() else "cpu", torch.float16
)

# Step 4: Generate answer
generate_ids = model.generate(**inputs, max_new_tokens=300)
output = processor.batch_decode(generate_ids, skip_special_tokens=True)[0]

# Step 5: Print answer
print("Answer:\n", output)
```
